```python
fname = input("Masukkan nama file : ")

try:
    fhand = open(fname)
except:
    print("File tidak dapat dibuka:", fname)
    exit()

counts = dict()

for line in fhand:
    line = line.rstrip()

    if line.startswith("From "):
        words = line.split()

        email = words[1]

        counts[email] = counts.get(email, 0) + 1

print(counts)
```

```python
fname = input("Masukkan nama file: ")

try:
    fhand = open(fname)
except:
    print("File tidak dapat dibuka:", fname)
    exit()

counts = dict()

for line in fhand:
    line = line.rstrip()

    if line.startswith("From "):
        words = line.split()

        email = words[1]

        domain = email.split("@")[1]

        counts[domain] = counts.get(domain, 0) + 1

print(counts)
```
