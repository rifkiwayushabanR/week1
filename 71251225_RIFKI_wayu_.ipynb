```python
n = int(input("Silahkan masukan angka yang anda mau\n>> "))
a = n
ulg = True

while ulg == True:
    if a % 2 == 0 and a != 2:
        a -= 1
    elif a % 3 == 0 and a != 3:
        a -= 1
    elif a % 5 == 0 and a != 5:
        a -= 1
    elif a % 7 == 0 and a != 7:
        a -= 1
    else:
        print(f"{a} adalah prima terdekat dari {n}")
        ulg = False
```
