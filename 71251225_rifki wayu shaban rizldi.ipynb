```python
with open('file1.txt', 'w') as f:
    f.write("Halo dunia\n")
    f.write("Saya belajar Python\n")
    f.write("Ini baris ketiga\n")
```

```python
with open('file2.txt', 'w') as f:
    f.write("Halo dunia\n")
    f.write("Saya belajar Java\n")
    f.write("Ini baris ketiga berbeda\n")
```

```python
with open('file1.txt', 'r') as f1, open('file2.txt', 'r') as f2:
    lines1 = f1.readlines()
    lines2 = f2.readlines()

max_len = max(len(lines1), len(lines2))

for i in range(max_len):
    line1 = lines1[i].rstrip('\n') if i < len(lines1) else "(tidak ada)"
    line2 = lines2[i].rstrip('\n') if i < len(lines2) else "(tidak ada)"
    
    if line1 != line2:
        print(f"Perbedaan pada baris ke-{i+1}:")
        print(f"File1: {line1}")
        print(f"File2: {line2}")
        print("-" * 30)
```

```python
with open('soal.txt', 'w') as f:
    f.write("1+1 = || 2\n")
    f.write("Bendera Indonesia? || Merah Putih\n")
    f.write("Kota gudeg adalah: || Yogyakarta\n")
    f.write("Komponen PC untuk penyimpanan file adalah... || harddisk\n")
    f.write("50 * 20 = || 1000\n")
```

```python
print("nama file: soal.txt\n")

with open('soal.txt', 'r') as f:
    for line in f:
        soal, kunci = line.strip().split("||")
        soal = soal.strip()
        kunci = kunci.strip().lower()
        
        print(soal)
        jawaban = input("Jawab: ").lower()
        
        if jawaban == kunci:
            print("Jawaban benar!")
        else:
            print("Jawaban salah!")
        print()
```
