```python
def perkalian(a, b):
    c = []
    e = 1
    while e < a:
        c.append(str(b))
        c.append("+")
        e += 1
    c.append(str(b))
    c.append("=")
    d = " ".join(c)
    return d

bil1 = int(input("Silahkan masukan bilangan pertama anda\n>> "))
bil2 = int(input("Silahkan masukan bilangan kedua anda\n>> "))

c = perkalian(bil1, bil2)
print(f"{c} {bil1 * bil2}")
```

```python
bil1 = int(input("Silahkan masukan bilangan pertama anda\n>> "))
bil2 = int(input("Silahkan masukan bilangan kedua anda\n>> "))
a = []
c = bil1

if c == bil2:
    print("Maaf angka yang anda masukan sama")
elif c < bil2:
    while c <= bil2:
        if c % 2 == 0:
            None
        else:
            a.append(str(c))
        c += 1
else:
    while c >= bil2:
        if c % 2 == 0:
            None
        else:
            a.append(str(c))
        c -= 1

b = " ".join(a)
print(f"Bilangan ganjil antara {bil1} dan {bil2} adalah \n>> {b}")
```

```python
jmlh = int(input("Silahkan masukan jumlah MatKul anda\n>> "))
a = 1
nile = 0

while a <= jmlh:
    c = input(f"Silahkan masukan nilai matkul ke-{a} anda\n>> ").lower()
    
    if c == "a":
        c = 4
    elif c == "b":
        c = 3
    elif c == "c":
        c = 2
    else:
        c = 1

    a += 1
    nile += c

ips = nile / jmlh

print(f"Nilai anda adalah {nile}\nIPS anda adalah {round(ips,2)}")
```
