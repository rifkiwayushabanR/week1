```python
import re

html = """
<div class="jadwal">
<h2>Pemrograman Lanjut (TI0123)</h2>
<p class="dosen">Dosen Pengampuh: Matahari Bakti Didanendya</p>
<p class="email">Email: matahari.bakti@ukdw.ac.id</p>
<p class="ruangan">Ruangan: GWM-301</p>
<p class="waktu">Senin 08.30 - 10.30 | 3 SKS</p>
</div>

<div class="jadwal">
<h2>Basis Data (TI0456)</h2>
<p class="dosen">Dosen Pengampuh: Antonius Rachmat C</p>
<p class="email">Email: anton.rachmat@ukdw.ac.id</p>
<p class="ruangan">Ruangan: GWM-205</p>
<p class="waktu">Selasa 10.30 - 12.30 | 2 SKS</p>
</div>

<div class="jadwal">
<h2>Kalkulus (MA0789)</h2>
<p class="dosen">Dosen Pengampuh: Yuan Lukito</p>
<p class="email">Email: yuan.lukito@ukdw.ac.id</p>
<p class="ruangan">Ruangan: PLATO-12</p>
<p class="waktu">Rabu 13.00 - 15.00 | 2 SKS</p>
</div>
"""

# ambil setiap blok jadwal
data = re.findall(r'<div class="jadwal">(.*?)</div>', html, re.DOTALL)

for item in data:
    # ambil data satu-satu
    matkul = re.search(r'<h2>(.*?)</h2>', item).group(1)
    dosen = re.search(r'Dosen Pengampuh:\s*(.*?)</p>', item).group(1)
    email = re.search(r'Email:\s*(.*?)</p>', item).group(1)
    ruangan = re.search(r'Ruangan:\s*(.*?)</p>', item).group(1)
    waktu = re.search(r'<p class="waktu">(.*?)</p>', item).group(1)

    # pecah waktu
    hari, jam_sks = waktu.split(" ", 1)
    jam, sks = jam_sks.split("|")
    jam = jam.strip()
    sks = sks.strip()

    # hitung durasi
    start, end = jam.split(" - ")

    def ke_menit(jam_str):
        j, m = map(int, jam_str.split("."))
        return j * 60 + m

    menit = ke_menit(end) - ke_menit(start)

    # output sesuai expected
    print(matkul)
    print(f"Dosen Pengampuh : {dosen}")
    print(f"Email Dosen     : {email}")
    print(f"Ruangan         : {ruangan}")
    print(f"Jadwal          : {hari} {jam} ({menit} menit)")
    print(f"SKS             : {sks}")
    print("-" * 40)
```
