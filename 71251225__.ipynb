```python
# kategori aplikasi
game = {'Mobile Legends', 'PUBG', 'Free Fire', 'TikTok'}
sosial = {'TikTok', 'Instagram', 'WhatsApp'}
edukasi = {'Ruangguru', 'Google Classroom', 'TikTok'}

# gabungkan semua aplikasi
semua = game | sosial | edukasi

print("Aplikasi yang hanya muncul di satu kategori:")
for app in semua:
    jumlah = 0

    if app in game:
        jumlah += 1
    if app in sosial:
        jumlah += 1
    if app in edukasi:
        jumlah += 1

    if jumlah == 1:
        print(app)

print("\nAplikasi yang muncul tepat di dua kategori:")
for app in semua:
    jumlah = 0

    if app in game:
        jumlah += 1
    if app in sosial:
        jumlah += 1
    if app in edukasi:
        jumlah += 1

    if jumlah == 2:
        print(app)
```

```python
data_list = [1, 2, 2, 3, 4]
print("List:", data_list)

data_set = set(data_list)
print("Menjadi Set:", data_set)

data_set2 = {'A', 'B', 'C'}
print("\nSet:", data_set2)

data_list2 = list(data_set2)
print("Menjadi List:", data_list2)

data_tuple = (10, 20, 20, 30)
print("\nTuple:", data_tuple)

data_set3 = set(data_tuple)
print("Menjadi Set:", data_set3)

data_set4 = {'Python', 'Java', 'C++'}
print("\nSet:", data_set4)

data_tuple2 = tuple(data_set4)
print("Menjadi Tuple:", data_tuple2)
```

```python
try:
    file1 = input("Masukkan nama file pertama: ")
    file2 = input("Masukkan nama file kedua: ")

    f1 = open(file1, 'r')
    f2 = open(file2, 'r')

    teks1 = f1.read().lower().split()
    teks2 = f2.read().lower().split()

    set1 = set(teks1)
    set2 = set(teks2)

    sama = set1 & set2

    print("\nKata yang muncul di kedua file:")
    for kata in sama:
        print(kata)

    f1.close()
    f2.close()

except:
    print("File tidak ditemukan atau tidak bisa dibaca")
```
