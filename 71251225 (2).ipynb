```python
bil1 = int(input("Silahkan masukan bilangan pertama anda\n>> "))
bil2 = int(input("Silahkan masukan bilangan kedua anda\n>> "))
a = []
c = bil1

if c == bil2:
    print("Maaf angka yang anda masukan sama")
elif c < bil2:
    while c <= bil2:
        if c % 2 == 0:
            None
        else:
            a.append(str(c))
        c += 1
else:
    while c >= bil2:
        if c % 2 == 0:
            None
        else:
            a.append(str(c))
        c -= 1

b = " ".join(a)
print(f"Bilangan ganjil antara {bil1} dan {bil2} adalah \n>> {b}")
```
